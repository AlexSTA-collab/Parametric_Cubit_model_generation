```python
import numpy as np
import matplotlib.pyplot as plt
rom matplotlib import rcParams
import matplotlib.ticker as ticker
import pickle
from collections import defaultdict
```

```python
# --- Global rcParams ---
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "axes.labelsize": 22,       # Axis labels fontsize
    "xtick.labelsize": 18,      # X tick labels fontsize
    "ytick.labelsize": 18,      # Y tick labels fontsize
    "xtick.direction": "in",    # X ticks inward
    "ytick.direction": "in",    # Y ticks inward
    "xtick.top": True,          # Show ticks on top
    "ytick.right": True,        # Show ticks on right
    #"axes.grid": True,          # Enable grid
    #"grid.linestyle": "--",     # Grid style
    #"grid.alpha": 0.4           # Grid transparency
})
```

```python
# --- Load pickled dictionary ---
with open("parametric_curves_creep.pkl", "rb") as f:
    data = pickle.load(f)
```

```python
# --- Display keys ---
print("Keys:")
for k in data.keys():
    print(k)
```

```python
# --- Group by the first key element (a_i) ---
grouped_by_a = defaultdict(list)
for (a, b), v in data.items()[:-9]:
    grouped_by_a[a].append(v)
```

```python
# --- Convert to numpy arrays ---
grouped_by_a = {a: np.array(vals) for a, vals in grouped_by_a.items()}
```

```python
# --- Display available a_i values ---
print("Unique a_i values:", list(grouped_by_a.keys()))
```

```python
plt.figure("Grouped by a_i")
for a, vals in grouped_by_a.items():
    plt.plot(vals[0,0,:], label=f"a={a}")
plt.xlabel("Index")
plt.ylabel("Values")
plt.legend()
plt.title("All values grouped by a_i")

# Move legend outside the plot, on the right
plt.legend(
    loc='center left',          # anchor the legend to the left center
    bbox_to_anchor=(1.05, 0.5), # place it just outside the right edge
    borderaxespad=0.
)

plt.tight_layout()  # adjust layout to make room for the legend
plt.show()
```

```python
# --- Group by first key element (a_i) ---
grouped_by_a = defaultdict(list)

for k, v in data.items():
    # make sure keys are tuples
    a = k[0] if isinstance(k, (tuple, list)) else k
    grouped_by_a[a].append(v)
```

```python
# --- Display available keys in original order ---
all_a = list(grouped_by_a.keys())
print("Available a_i keys (in dictionary order):")
for i, a in enumerate(all_a):
    print(f"[{i}] a = {a},  number of entries = {len(grouped_by_a[a])}")
```

```python
# --- Choose which indices to plot ---
selected_indices = [0, 3, 5]  # choose which a_i you want
selected_keys = [all_a[i] for i in selected_indices]
for i, a in enumerate(all_a):
    print(f"[{i}] a = {a},  number of entries = {len(grouped_by_a[a])}")
```

```python
plt.figure("Grouped by a_i (selected)")
for a in selected_keys:
    vals = np.array(grouped_by_a[a])  # convert list -> array, shape (3, 42)
    print(vals.shape)
    #for j in range(vals.shape[0]):
    plt.plot(vals[0,0,:], label=f"a={a}")

plt.xlabel("Index")
plt.ylabel("Value")
plt.legend()
plt.title("Selected curves grouped by a_i")
plt.tight_layout()
plt.show()
```
